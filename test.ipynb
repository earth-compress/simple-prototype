```python
from ec_prototype.datasets import DATASETS, DatasetKey
import matplotlib.pyplot as plt
```

```python
# Select and download the data
dataset = DATASETS[DatasetKey.GEORVQ]
dataset.download()
```

```python
# Decode the full data (with GPU)
grids = dataset.decode(device="mps", quality="8-8")
bands = dataset.bands
bands
```

```python
plt.imshow(grids["60m"][0])
plt.colorbar()
```

```python
# Different quality
grids2 = dataset.decode(device="mps", quality="8-0")
plt.imshow(grids2["60m"][0])
plt.colorbar()
```

```python
# Plot difference of the quality levels
diff = grids["60m"][0] - grids2["60m"][0]
plt.imshow(diff)
plt.colorbar()
```
