````python
# Import the required libraries

import matplotlib.pyplot as plt

from ec_prototype.datasets import DATASETS, DatasetKey
````

## GeoRVQ Example

The following variables can be updated for testing purposes:

- **Quality level**: This controls how much information is used when reconstructing the compressed image.
To update this, set the `quality` to one of the following: "8-0", "8-4", "8-8"

- **Spatial resolution**: Describes how much ground area each pixel represents.
To update this, set the `spatial_resolution` to one of the following: "10m", "20m" and "60m"

- **Compute device**: Determines which hardware is used to decode the data.
To update this, add `device` in the decode method to one of the following: 'cuda', 'cpu', 'mps'.
Note that the default is set to None, this automatically selects an available compute device.

See [here](https://huggingface.co/datasets/earth-compress/EC-prototype-S2L1C) for further information on how these formats work.

````python
# 1. Select and download the data in GeoRVQ format
georvq_dataset = DATASETS[DatasetKey.GEORVQ]
georvq_dataset.download()
````

````python
# 2. Decode the full data

# Set the variables based on the quality and spatial resolution to test
quality_1 = "8-8"
spatial_resolution_1 = "60m"

grids = georvq_dataset.decode(quality=quality_1)
bands = georvq_dataset.bands
plt.imshow(grids[spatial_resolution_1][0])
plt.title("Display the downloaded image")
plt.colorbar()
````

````python
# 3. Compare the different quality levels
figure, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].imshow(grids[spatial_resolution_1][0])
axes[0].set_title(f"GeoRVQ Quality: {quality_1}")

# Download a different quality level
quality_2 = "8-0"
spatial_resolution_2 = "60m"
grids2 = georvq_dataset.decode(quality=quality_2)
axes[1].imshow(grids2[spatial_resolution_2][0])
axes[1].set_title(f"GeoRVQ Quality: {quality_2}")

# Plot the difference of the quality levels
diff = grids[spatial_resolution_1][0] - grids2[spatial_resolution_2][0]
axes[3].imshow(grids[spatial_resolution_1][0])
axes[3].set_title(f"Difference between {quality_1} and {quality_2}")

plt.show()
````

## JPEG XL Example

The following variables can be updated for testing purposes:

- **Quality level**: This controls how much information is used when reconstructing the compressed image.
To update this, set the `quality` to one of the following: d-0.094, d-0.17, d-0.33, d-0.63, d-1.2

To update this, set the `grids["60m"]` to one of the following: "10m", "20m" and "60m"

See [here](https://huggingface.co/datasets/earth-compress/EC-prototype-S2L1C) for further information on how these formats work.

````python
# 1. Select and download the data in JPEG XL format
jpegxl_dataset = DATASETS[DatasetKey.JPEGXL]
jpegxl_dataset.download()
````

````python
# 2. Decode the full data

# Set the variables based on the quality and spatial resolution to test
quality_1 = "d-1.2"
spatial_resolution_1 = "60m"

grids = jpegxl_dataset.decode(quality=quality_1)
bands = jpegxl_dataset.bands
plt.imshow(grids[spatial_resolution_1][0])
plt.title(f"JPEG XL Quality: {quality_1}")

plt.colorbar()
````

````python
# 3. Compare the different quality levels
figure, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].imshow(grids[spatial_resolution_1][0])
axes[0].set_title(f"JPEG XL Quality: {quality_1}")

# Download a different quality level
quality_2 = "d-0.17"
spatial_resolution_2 = "60m"
axes[1].imshow(grids2[spatial_resolution_2][0])
axes[1].set_title(f"JPEG XL Quality: {quality_2}")

# Plot the difference of the quality levels
diff = grids[spatial_resolution_1][0] - grids2[spatial_resolution_2][0]
axes[3].imshow(grids[spatial_resolution_1][0])
axes[3].set_title(f"Difference between {quality_1} and {quality_2}")

plt.show()
````
