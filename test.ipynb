```python
import matplotlib.pyplot as plt

from ec_prototype.datasets import DATASETS, DatasetKey
```

## GeoRVQ example

```python
# Select and download the data in Georvq format
georvq_dataset = DATASETS[DatasetKey.GEORVQ]
georvq_dataset.download()
```

```python
# Decode the full data
grids = georvq_dataset.decode(device="mps", quality="8-8")
bands = georvq_dataset.bands
plt.imshow(grids["60m"][0])
plt.title("Display the downloaded image")
plt.colorbar()
```

```python
# Decode the data at a different quality level.
# Note that this doesn't require a re-download.
grids2 = georvq_dataset.decode(device="mps", quality="8-0")
plt.imshow(grids2["60m"][0])
plt.colorbar()
```

```python
# Plot the difference of the quality levels
diff = grids["60m"][0] - grids2["60m"][0]
plt.imshow(diff)
plt.colorbar()
```

```python
# Compare the different quality levels
figure, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].imshow(grids["60m"][0])
axes[0].set_title("GeoRVQ Quality: 8-8")

axes[1].imshow(grids2["60m"][0])
axes[1].set_title("GeoRVQ Quality: 8-0")

plt.show()
```
