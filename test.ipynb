# Earth Compress notebook

This notebook showcases downloading and decoding datasets compressed under different sample algorithms.

There are two examples in use, [GeoRVQ](https://github.com/earth-compress/georvq) and [JPEG XL](https://jpegxl.info/).

Each example provides instructions on what variables can be adjusted to produce different image qualities.

The sample Sentinel-2 files used in this repository can be found [here](https://huggingface.co/datasets/earth-compress/EC-prototype-S2L1C)

## Install dependencies

Requires [uv](https://docs.astral.sh/uv/getting-started/installation/) to be installed.

To install necessary dependencies run: 
```
uv sync --group demo
```
This will create a virtual environment which should then be used for the kernel of the notebook.


````python
# Import the required libraries

import matplotlib.pyplot as plt

from ec_prototype.datasets import DATASETS, DatasetKey
````

## GeoRVQ Example

The following GeoRVQ example variables can be updated for testing purposes:

- **Quality level**: This controls how much information is used when reconstructing the compressed image.
To update this, set the `quality` to one of the following: "8-0", "8-4", "8-8"

- **Spatial resolution**: Describes how much ground area each pixel represents.
To update this, set the `grids["60m"]` to one of the following: "10m", "20m" and "60m"

See [here](https://huggingface.co/datasets/earth-compress/EC-prototype-S2L1C) for further information on how these formats work.

````python
# 1. Select and download the data in GeoRVQ format
georvq_dataset = DATASETS[DatasetKey.GEORVQ]
georvq_dataset.download()
````

````python
# 2. Decode the full data
grids = georvq_dataset.decode(device="mps", quality="8-8")
bands = georvq_dataset.bands
plt.imshow(grids["60m"][0])
plt.title("Display the downloaded image")
plt.colorbar()
````

````python
# 3. Decode the data at a different quality level
# Note that this doesn't require a re-download.
grids2 = georvq_dataset.decode(device="mps", quality="8-0")
plt.imshow(grids2["60m"][0])
plt.colorbar()
````

````python
# 4. Plot the difference of the quality levels
diff = grids["60m"][0] - grids2["60m"][0]
plt.imshow(diff)
plt.colorbar()
````

````python
# 5. Compare the different quality levels
figure, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].imshow(grids["60m"][0])
axes[0].set_title("GeoRVQ Quality: 8-8")

axes[1].imshow(grids2["60m"][0])
axes[1].set_title("GeoRVQ Quality: 8-0")

plt.show()
````
